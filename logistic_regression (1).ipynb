#**Implementation of Logistic Regression using (Scikit-Learn)**

### 1. Import Scikit-Learn Modules
We begin by importing the necessary tools for model building, splitting data, and evaluating performance.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
```

```python
df.head()
```

### 2. Define Features and Target
Here we specify which columns will be used to make predictions (Features) and which column we want to predict (Target).

**Separate the features and target**

```python
features = ['Age', 'Gender', 'Blood Type', 'Medical Condition', 'Admission Type']
target = 'Test Results'

X = df[features]
Y = df[target]
```

### Encoding Categorical Variables
**Since machine learning models require numerical input, we will use `pd.get_dummies()` to convert text columns into numerical format**

```python
# Select categorical features to encode
features = ['Gender', 'Blood Type', 'Medical Condition', 'Admission Type']

# Create dummy variables
X_encoded = pd.get_dummies(X, columns=features, drop_first=True)

# Re-split the data with the encoded features
X_train, X_test, Y_train, y_test = train_test_split(
    X_encoded, Y,
    test_size=0.2,
    random_state=50
)

print("Encoded features preview:")
display(X_encoded.head())
```

### 3. Initialize the Model
We create an instance of the Logistic Regression model.

```python
model=LogisticRegression()
```

### 4. Train the Model
We feed the training data into the model so it can learn the relationships between the features and the test results.

```python
model.fit(X_train,Y_train)
```

### 5. Generate Predictions
Using the unseen test data, we ask the model to predict the test results.

```python
y_pred = model.predict(X_test)
```

### 6. View Raw Predictions
We print the array of predicted labels to verify the output format.

```python
print(y_pred)
```

### Comparison of Actual vs Predicted Values
To see exactly how the model is predicting, let's look at the first 10 rows of our test set alongside the model's guesses.

```python
# Create a DataFrame to compare actual vs predicted
comparison_df = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})

# Show the first 10 comparisons
display(comparison_df.head(10))
```

### 7. Calculate Accuracy
We measure how often the model's predictions match the actual results in the test set.

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

### 8. Compute Confusion Matrix
This shows the count of true positives, true negatives, and misclassifications for each category.

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    xticklabels=['Pass', 'Fail'],
    yticklabels=['Pass', 'Fail']
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix Heatmap")
plt.show()
```

### 9. Calculate Precision
Precision measures the accuracy of the positive predictions.

```python
precision = precision_score(y_test, y_pred, average='weighted')
print(f"Precision: {round(precision, 2)}")
```

### 10. Calculate Recall
Recall measures the ability of the model to find all the positive cases.

```python
recall= recall_score(y_test, y_pred, average='weighted')
print(f"Precision: {round(recall, 2)}")
```

### 11. Calculate F1-Score
The F1-Score provides a balance between Precision and Recall.

```python
f1= f1_score(y_test, y_pred, average='weighted')
print(f"Precision: {round(f1, 2)}")
```

### 12. Predict for a New Patient
 we use our trained model to predict the result for an entirely new patient entry.

```python


# 1. Define new patient data
new_patient_data = pd.DataFrame({
    'Age': [45],
    'Gender': ['male'],
    'Blood Type': ['A+'],
    'Medical Condition': ['diabetes'],
    'Admission Type': ['urgent']
})

# 2. Encode the new data
new_patient_encoded = pd.get_dummies(new_patient_data)


# This adds missing columns as 0 and removes extra ones
new_patient_final = new_patient_encoded.reindex(columns=X_train.columns, fill_value=0)

# 4. Predict!
prediction = model.predict(new_patient_final)
print(f"The predicted Test Result is: {prediction[0]}")
```
